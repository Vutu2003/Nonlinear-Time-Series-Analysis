#  Standardized NTSA Research Framework
**Objective:** Reproduce Costa et al. (2005) Multiscale Entropy (MSE) physiological experiments, laying the foundation for a scalable Non-Linear Time Series Analysis (NTSA) pipeline (Entropy $\rightarrow$ Chaos $\rightarrow$ Fractal $\rightarrow$ Recurrence).

##  1. Architecture Design (Decoupled Layers)
This framework strictly separates the **Data Layer** from the **Analysis Layer**. The algorithm (MSE) is treated as a swappable "black-box engine".

*   **[Experiment Layer]** (Notebook Cells): Orchestrates the data flow and controls variables for specific hypotheses.
*   **[Analysis Utilities]**: Post-processing metrics (Slope, AUC, Max/Mean, Crossover extraction).
*   **[Analysis Engine]**: The core mathematical algorithm (Currently: `MSE`. Future: `DFA`, `RQA`).
*   **[Data Layer]**: Reusable pipeline for preprocessing, segmenting, and perturbing physiological signals.

##  2. Core Modules (Preprocessing Utilities)
1.  **Data Acquisition (`download_rr`, `extract_rr`)**: PhysioNet DB interaction and RR interval parsing.
2.  **Segmentation (`extract_sleep`, `extract_wake`, `extract_window`)**: Standardized data cropping (e.g., $N = 2 \times 10^4$).
3.  **Perturbation (`inject_noise`, `inject_artifacts`, `shuffle_series`)**: Stress-testing algorithms via Gaussian noise, outliers, and Surrogate Data generation.
4.  **Parameter Estimation (`estimate_clean_sd`, `estimate_r`)**: Robust tolerance $r$ calculation filtering out extreme ectopics.
5.  **Debug Tools (`coarse_grain_preview`)**: Visual inspection of signal transformation across scales.

##  3. Experiment Roadmap (Costa 2005)
*   **Experiment 1 (Group Dynamics - Fig 5):** Healthy (NSR) vs. CHF vs. AF. Identifies the "crossover" phenomenon.
*   **Experiment 2 (Age & Circadian - Fig 6 & 7):** Wake vs. Sleep segments across Young/Old/CHF groups.
*   **Experiment 3 (Clinical Feature Extraction - Fig 8):** Scatter plot of $\text{Slope}_{1-5}$ vs. $\text{Slope}_{6-20}$ for classification.
*   **Experiment 4 (Noise Resilience - Fig 16):** Injecting 10% & 20% Gaussian noise to observe curve convergence at larger scales.
*   **Experiment 5 (Artifact Tolerance - Fig 18):** Injecting extreme outliers to validate the Costa filtering trick ($r$ computed from clean SD).
*   *(Bonus)* **Surrogate Data Test:** Shuffling RR series to destroy non-linear correlations (Preparation for future Chaos studies).
